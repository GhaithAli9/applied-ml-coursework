```python
import requests 
import matplotlib.pyplot as plt
```

```python
url="https://raw.githubusercontent.com/aalanwar/Logical-Zonotope/refs/heads/main/README.md"
response = requests.get(url)
with open("readme.md", 'w', encoding='utf-8') as f:
    f.write(response.text)
```

```python
words=response.text.lower().split()
print("the split function produced :",words)
```

```python
import pandas as pd 
s= pd.Series(words)
counts=s.value_counts()
print(counts.head(10))
```

```python
stopwords = {'the', 'a', 'an', 'be', 'of', 'and', 'to', 'in', 'is', 'for'}
filtered_words = [w for w in words if w.isalpha() and w not in stopwords]
filtered_words[:20]
```

```python
s=pd.Series(filtered_words)
counts=s.value_counts()
```

```python
counts.head(10).plot(kind='bar')
plt.title('Top 10 most frequent words')
plt.xlabel('Words')
plt.ylabel('Freqiency')
plt.show()
```
